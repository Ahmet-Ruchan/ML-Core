```python
# independent (input) and dependent (output) features
X = df[['Study Hours', 'Sleep Hours', 'Attendance Rate', 'Social Media Hours']] # 2D array for multiple features
y = df['Exam Score'] # 1D array

#X = df.iloc[:, :-1] # all rows, all columns except the last column
#y = df.iloc[:, -1] # all rows, only the last column
```

```python
X.head()
```

```python
y.head()
```

```python
# train - test split
from sklearn.model_selection import train_test_split

X_train, X_test, y_tarin, y_test = train_test_split(X, y, test_size=0.25, random_state=42)
```

```python
# Standardize the data
from sklearn.preprocessing import StandardScaler

scaler = StandardScaler()
X_tarin = scaler.fit_transform(X_train) # fit and transform the training data
X_test = scaler.transform(X_test) # only transform the test data
```

```python
X_tarin
```

```python
# Training the Model with Linear Regression

from sklearn.linear_model import LinearRegression

regression = LinearRegression()
regression.fit(X_tarin, y_tarin) # training the model with training data X_train girdileri ve y_train çıktılarını veriyor bunu model öğreniyor
```

```python
X_test[0] # first test sample
```

```python
df.iloc[0] # first sample in the original dataframe
```

```python
# Predicting the Exam Score for the first test sample
new_student = [[5, 7, 90, 3]] # 5 study hours, 7 sleep hours, 90% attendance rate, 3 hours on social media
new_Student_scaled = scaler.transform(new_student) # scale the new student data
predicted_score = regression.predict(new_Student_scaled)
predicted_score
```

```python
# Evaluating the Model
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score

y_pred = regression.predict(X_test) # predict the exam scores for the test set
y_pred
```

```python
mae = mean_absolute_error(y_test, y_pred) # y_test are the true values, y_pred are the predicted values
mse = mean_squared_error(y_test, y_pred)
rmse = np.sqrt(mse)
r2 = r2_score(y_test, y_pred)

print("Mean Absolute Error (MAE):", mae)
print("Mean Squared Error (MSE):", mse)
print("Root Mean Squared Error (RMSE):", rmse)
print("R-squared (R2 ):", r2)
```

```python
# Visualizing the Actual vs Predicted Exam Scores

plt.scatter(y_test, y_pred, color='blue')
plt.xlabel('Actual Exam Score')
plt.ylabel('Predicted Exam Score')
plt.title('Actual vs Predicted Exam Score')
#plt.plot([0, 100], [0, 100], color='red', linestyle='--') # diagonal line
plt.show()
```

```python
residuals = y_test - y_pred # residuals = actual - predicted
sns.displot(residuals, kde=True) # distribution of residuals kde=True to show the kernel density estimate
plt.title('Distribution of Residuals')
plt.xlabel('Residuals')
plt.ylabel('Frequency')
plt.show()
```
