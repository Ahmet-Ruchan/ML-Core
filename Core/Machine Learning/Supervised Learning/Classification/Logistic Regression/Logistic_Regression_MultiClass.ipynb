```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from sklearn import datasets
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import confusion_matrix, accuracy_score, classification_report
from sklearn.model_selection import GridSearchCV, StratifiedKFold
```

```python
df = pd.read_csv("/Users/ruch/Desktop/ML-Engineer/Core/Machine Learning/Data/7-cyber_attack_data.csv")
df.head()
```

```python
df.columns
```

```python
df.info()
```

```python
X = df.drop("attack_type", axis=1)
y = df["attack_type"]
```

```python
from sklearn.model_selection import train_test_split
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.3, random_state=42)
```

```python
from sklearn.linear_model import LogisticRegression
logistic_reg = LogisticRegression()
logistic_reg.fit(X_train, y_train)
y_pred = logistic_reg.predict(X_test)
```

```python
y_pred
```

```python
from sklearn.metrics import confusion_matrix, classification_report, accuracy_score
```

```python
score = accuracy_score(y_pred, y_test)
print("score:" , score)
confusion_matrix_score = confusion_matrix(y_pred, y_test)
print("confusion matrix: \n", confusion_matrix_score)
classification_report_score = classification_report(y_pred, y_test)
print("classification report: \n", classification_report_score)
```

```python
# one vs rest
```

```python
from sklearn.multiclass import OneVsOneClassifier, OneVsRestClassifier
```

```python
one_vs_one_model = OneVsOneClassifier(LogisticRegression())
one_vs_one_model.fit(X_train, y_train)
y_pred_ovo = one_vs_one_model.predict(X_test)
```

```python
score = accuracy_score(y_pred_ovo, y_test)
print("ovo score:" , score)
confusion_matrix_score = confusion_matrix(y_pred_ovo, y_test)
print("ovo confusion matrix: \n", confusion_matrix_score)
classification_report_score = classification_report(y_pred_ovo, y_test)
print("ovo classification report: \n", classification_report_score)
```

```python
one_vs_resrt_model = OneVsRestClassifier(LogisticRegression())
one_vs_resrt_model.fit(X_train, y_train)
y_pred_ovr = one_vs_resrt_model.predict(X_test)
```

```python
score = accuracy_score(y_pred_ovr, y_test)
print("ovr score:" , score)
confusion_matrix_score = confusion_matrix(y_pred_ovr, y_test)
print("ovr confusion matrix: \n", confusion_matrix_score)
classification_report_score = classification_report(y_pred_ovr, y_test)
print("ovr classification report: \n", classification_report_score)
```
