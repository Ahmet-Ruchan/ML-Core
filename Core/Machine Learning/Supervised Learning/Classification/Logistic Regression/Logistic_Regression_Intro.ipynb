```python
import pandas as pd
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
df = pd.read_csv("/Core/Machine Learning/Data/6-bank_customers.csv")
```

```python
df.head()
```

```python
df.columns
```

```python
df.info()
```

```python
df.describe()
```

```python
X = df.drop("subscribed", axis=1)
y = df["subscribed"]
```

```python
from sklearn.model_selection import train_test_split
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.3, random_state=42)
```

```python
from sklearn.linear_model import LogisticRegression
logistic_reg = LogisticRegression()
logistic_reg.fit(X_train, y_train)
y_pred = logistic_reg.predict(X_test)
```

```python
y_pred
```

```python
from sklearn.metrics import confusion_matrix, accuracy_score, classification_report
```

```python
score = accuracy_score(y_pred, y_test)
print("score:" , score)
```

```python
cr = classification_report(y_pred, y_test)
print("classification report: \n", cr)
```

```python
cm = confusion_matrix(y_pred, y_test)
print("confusion matrix: \n", cm)
```

```python
# Hyperparameter Tuning
```

```python
model = LogisticRegression()
```

```python
penalty = ['l1', 'l2', 'elasticnet', 'none']
c_values = [100, 10, 1, 0.1, 0.01]
solver = ['lbfgs', 'newton-cg', 'liblinear', 'sag', 'saga', 'newton-cholesky']
```

```python
params = dict(
    penalty=penalty,
    C=c_values,
    solver=solver
)
```

```python
params
```

```python
# grid search cv
from sklearn.model_selection import GridSearchCV, StratifiedKFold
```

```python
cv = StratifiedKFold(n_splits=5, shuffle=True, random_state=42)
```

```python
grid = GridSearchCV(estimator=model, param_grid=params, cv=cv, scoring='accuracy', n_jobs=-1, verbose=1)
```

```python
grid
```

```python
grid.fit(X_train, y_train)
```

```python
grid.best_params_
```

```python
grid.best_score_
```

```python
y_pred = grid.predict(X_test)
```

```python
score = accuracy_score(y_pred, y_test)
print("score:" , score)

cr = classification_report(y_pred, y_test)
print("classification report: \n", cr)

cm = confusion_matrix(y_pred, y_test)
print("confusion matrix: \n", cm)
```

```python
# random search cv
from sklearn.model_selection import RandomizedSearchCV
```

```python
model = LogisticRegression()
```

```python
randomcv = RandomizedSearchCV(estimator=model, param_distributions=params, n_iter=10, cv=cv, scoring='accuracy', n_jobs=-1, verbose=1, random_state=42)
```

```python
randomcv
```

```python
randomcv.fit(X_train, y_train)
```

```python
randomcv.best_params_
```

```python
randomcv.best_score_
```

```python
y_pred = randomcv.predict(X_test)
```

```python
score = accuracy_score(y_pred, y_test)
print("score:" , score)

cr = classification_report(y_pred, y_test)
print("classification report: \n", cr)

cm = confusion_matrix(y_pred, y_test)
print("confusion matrix: \n", cm)
```
