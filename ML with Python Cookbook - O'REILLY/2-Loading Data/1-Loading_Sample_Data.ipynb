# Creating a Simulated Dataset

```python
# When we want a dataset designed to be used with linear regression, we can use the make_regression function from sklearn.datasets

from sklearn.datasets import make_regression

features, target, coefficients = make_regression(n_samples=100, # 100 samples
                                                 n_features=3, # Three features
                                                 n_informative=3, # All features are informative
                                                 n_targets=1, # Single target variable
                                                 noise=0.0, # No noise added to the data
                                                 coef=True, # Return the true coefficients
                                                 random_state=1) # Setting a random state ensures reproducibility

print("Feature Matrix:\n", features[:3])
print("Target Vector:\n", target[:3])
print("Coefficients:\n", coefficients)

import matplotlib.pyplot as plt

plt.scatter(features[:, 0], target)
plt.xlabel("Feature 1")
plt.ylabel("Feature 2")
plt.title("Simulated Regression Data")
plt.show()
```

```python
# When we want a dataset designed to be used with classification algorithms, we can use the make_classification function from sklearn.datasets

features, target = datasets.make_classification(n_samples=100, # 100 samples
                                                n_features=3, # 3 features
                                                n_informative=3, # Three informative features
                                                n_redundant=0, # 0 redundant feature
                                                n_classes=2, # Two classes
                                                weights=[.25, .75], # Class distribution
                                                random_state=1,) # Setting a random state ensures reproducibility

print("Feature Matrix:\n", features[:3])
print("Target Vector:\n", target[:3])

import matplotlib.pyplot as plt

plt.scatter(features[:, 0], features[:, 1], c=target, cmap='viridis')
plt.xlabel("Feature 1")
plt.ylabel("Feature 2")
plt.title("Simulated Classification Data")
plt.show()
```

```python
# When we want a dataset designed to be used with clustering algorithms, we can use the make_blobs function from sklearn.datasets

from sklearn.datasets import make_blobs

features, target = make_blobs(n_samples=100, # 100 samples
                              n_features=2, # Two features
                              centers=3, # Three centers
                              cluster_std=0.5, # Standard deviation of clusters
                              shuffle=True, # Shuffle the samples
                              random_state=1) # Setting a random state ensures reproducibility

print("Feature Matrix:\n", features[:3])
print("Target Vector:\n", target[:3])

import matplotlib.pyplot as plt

plt.scatter(features[:, 0], features[:, 1], c=target, cmap='viridis')
plt.xlabel("Feature 1")
plt.ylabel("Feature 2")
plt.title("Simulated Clustering Data")
plt.show()
```

```python
# What is the difference between make_blobs and make_classification and make_regression?

# make_blobs is used for generating datasets for clustering tasks, where the goal is to group similar data points together based on their features. The generated data points are grouped around specified centers, making it suitable for testing clustering algorithms.

# make_classification is used for generating datasets for classification tasks, where the goal is to assign data points to predefined classes based on their features. The generated data points are designed to be separable into distinct classes, making it suitable for testing classification algorithms.

# make_regression is used for generating datasets for regression tasks, where the goal is to predict a continuous target variable based on input features. The generated data points have a linear relationship with the target variable, making it suitable for testing regression algorithms
```

Üçü de scikit-learn içinde sentetik veri üretmek için kullanılıyor ama amaçları farklı:

⸻

🔹 *make_blobs*
	•	Kullanım amacı: Kümeleme (clustering) algoritmalarını denemek.
	•	Nasıl çalışır: Belirli sayıda küme (cluster) oluşturur. Her kümenin merkezi ve dağılımı vardır.
	•	Çıktı: Özellikler (X) ve küme etiketleri (y).
	•	👉 Daha çok K-Means, DBSCAN gibi *Unsupervised Learning* gözetimsiz öğrenme algoritmaları için.

⸻

🔹 *make_classification*
	•	Kullanım amacı: Sınıflandırma (classification) problemleri için gerçekçi yapıda veri üretmek.
	•	Nasıl çalışır: Belirlenen sayıda sınıf, bilgilendirici (informative), gereksiz (redundant) ve gürültülü özellikler üretir.
	•	Çıktı: Özellikler (X) ve sınıf etiketleri (y).
	•	👉 Daha çok lojistik regresyon, karar ağaçları, SVM gibi *Supervised Learning* gözetimli öğrenme algoritmalarını denemek için.

⸻

🔹 *make_regression*
	•	Kullanım amacı: Regresyon (sürekli değer tahmini) problemleri için veri üretmek.
	•	Nasıl çalışır: Özellikler (X) ile sürekli hedef değerler (y) arasında doğrusal ilişki kurar, istenirse gürültü ekler.
	•	Çıktı: Özellikler (X) ve sürekli hedef (y).
	•	👉 Daha çok linear regression, ridge, lasso, SVR gibi algoritmalar için.

⸻

✅ Kısacası:
	•	make_blobs → Kümeleme (cluster) testleri.
	•	make_classification → Sınıflandırma testleri.
	•	make_regression → Regresyon testleri.
